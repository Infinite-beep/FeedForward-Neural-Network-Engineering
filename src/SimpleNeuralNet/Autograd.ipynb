```python
import torch
```

```python
a = torch.tensor(2.0)
b = torch.tensor(2.0)
```

Each Tensor contains the following attributes
1. Data which store the data i.e for this example it stores 2.0
2. grad : stores the calculated gradient
3. grad_fn : Points to a node in the backward graph
4. is_leaf : Checks whether a particular node tensor is a leaf. Essential a leaf node that does not track gradients and node that track gradient and intialized by user are leaf nodes
5. requires_grad

```python
c = a * b
print(c)

print(a.requires_grad)
print(b.requires_grad)
```

```python
# Create a tensor with requires_grad = True
d = torch.tensor(5., requires_grad = True)
e = torch.tensor(5)

f = d + e
```

```python
print(f'd: {d.requires_grad}')
print(f'e: {d.requires_grad}')
print(f'f: {d.requires_grad}')

# Backward graph of f
print(f.grad_fn)
```

```python
f.grad_fn.next_functions
```

```python
# Backprop
f.backward()
print(d.grad)
```

```python
#Another Example
s = torch.tensor(2.0, requires_grad = True)
r = torch.tensor(3.0, requires_grad = True)

u = s * r

v = torch.tensor(4.0, requires_grad = True)

w = u * v
```

```python
#requires_grad attribute
print(s.requires_grad)
print(r.requires_grad)
print(u.requires_grad)
print(v.requires_grad)
print(w.requires_grad)
```

```python
print(u.grad_fn)
print(w.grad_fn)
```

```python
from torchviz import make_dot
```

```python
w.backward()
```

```python

dot = make_dot(w, params={"s": s, "r": r, "v": v})
dot.render(filename="Autograd", format="jpg")
```

```python
# Gradient
print(s.grad)
print(r.grad)
print(u.grad)
print(v.grad)
print(w.grad)
```
